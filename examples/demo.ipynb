# coilcal demo
Typical coil: 1.0 mm wire, 2.0 mm pitch, 150 mm former.

```
import matplotlib.pyplot as plt
import numpy as np

import coilcal as cc

a = cc.SolenoidCoil(wire_d=1.0e-3, pitch=2.0e-3, former_d=0.150, n_turns=10)
b = cc.SolenoidCoil(wire_d=1.0e-3, pitch=2.0e-3, former_d=0.150, n_turns=5, position=[0, 0, 0.065])
print(cc.check_quasistatic(a, 5e6))
```

```
m = cc.mutual_inductance(a, b)
print(f"M = {m * 1e6:.4g} uH, k = {cc.coupling_coefficient(a, b, m=m):.4f}")
print(f"|EMF| in B for 1 A @ 5 MHz in A: {abs(cc.induced_emf(a, b, 1.0, 5e6, m=m)):.4g} V")
```

```
z = np.linspace(-0.15, 0.15, 301)
pts = np.column_stack([0 * z, 0 * z, z])
plt.plot(z * 1e3, cc.field_B(a, pts)[:, 2] * 1e6, label="rings (magpylib)")
plt.plot(z * 1e3, cc.field_B(a, pts, model="helix")[:, 2] * 1e6, "--", label="helix (magpylib)")
plt.xlabel("z [mm]")
plt.ylabel("B_z per ampere [uT/A]")
plt.legend();
```
